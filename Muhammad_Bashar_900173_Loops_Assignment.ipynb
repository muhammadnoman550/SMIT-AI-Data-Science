### Example:

* Output: 5

```python
number = abs(int(input("Enter a number: ")))
count = 0

if number == 0:
    count = 1
else:
    while number > 0:
        number //= 10
        count += 1

print("Number of digits:", count)
```

## Write a program to calculate the factorial of a number.
### Example:

* Input : 5
* Output 120

```python
number = int(input("Enter a non-negative number: "))
factorial = 1

if number < 0:
    print("Factorial is not defined for negative numbers.")
else:
    for value in range(1, number + 1):
        factorial *= value
    print("Factorial:", factorial)
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []

for student in range(1, 6):
    raw_marks = input(f"Enter marks for student {student} (-1 to stop): ").strip()

    if raw_marks == "-1":
        break

    try:
        mark = float(raw_marks)
    except ValueError:
        print("Invalid marks")
        continue

    if mark < 0 or mark > 100:
        print("Invalid marks")
        continue

    marks.append(mark)

print("Valid Marks:", marks)
if marks:
    print("Average =", sum(marks) / len(marks))
else:
    print("No valid marks were entered.")
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
available_rooms = [101, 102, 103, 104, 105]
booked_rooms = []

while available_rooms:
    print("Available rooms:", available_rooms)
    try:
        room = int(input("Choose a room: "))
    except ValueError:
        print("Please enter a room number.")
        continue

    if room not in available_rooms:
        print("Room not available.")
        continue

    available_rooms.remove(room)
    booked_rooms.append(room)
    print(f"Room {room} booked successfully!")

print("All rooms have been booked.")
print("Booked rooms:", booked_rooms)
```

## Car Rental System

### Goal:
    Create a car rental booking system.

### Steps:

    Cars = ["Civic", "Corolla", "City", "Alto"]
    
    Use a for loop to let multiple users book.
    
    If user enters "stop" → break.
    
    If car not in list → continue (invalid).
    
    Add valid bookings to booked_cars list.

### Example:

    Enter car name: Civic  
    Enter car name: Alto  
    Enter car name: Tesla → Not available  
    Enter car name: stop  
    Booked Cars: ['Civic', 'Alto']

```python
cars = ["Civic", "Corolla", "City", "Alto"]
booked_cars = []

for _ in range(100):
    car = input("Enter car name (or 'stop' to finish): ").strip()

    if car.lower() == "stop":
        break

    matching_car = next((name for name in cars if name.lower() == car.lower()), None)
    if matching_car is None:
        print("Not available")
        continue

    if matching_car in booked_cars:
        print("This car is already booked.")
        continue

    booked_cars.append(matching_car)
    print(f"{matching_car} booked successfully.")

print("Booked Cars:", booked_cars)
```

## Electricity Bill Tracker

### Example:

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0

while True:
    print("\n--- Banking Menu ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    choice = input("Choose an option: ").strip()

    if choice == "1":
        print("Account is ready to use.")
    elif choice == "2":
        try:
            amount = float(input("Enter deposit amount: "))
            if amount <= 0:
                print("Amount must be positive.")
            else:
                balance += amount
                print("Deposit successful.")
        except ValueError:
            print("Enter a valid amount.")
    elif choice == "3":
        try:
            amount = float(input("Enter withdrawal amount: "))
            if amount <= 0:
                print("Amount must be positive.")
            elif amount > balance:
                print("Insufficient balance.")
            else:
                balance -= amount
                print("Withdrawal successful.")
        except ValueError:
            print("Enter a valid amount.")
    elif choice == "4":
        print(f"Current balance: {balance:.2f}")
    elif choice == "5":
        print("Thank you for using the banking system.")
        break
    else:
        print("Invalid option. Please choose 1 to 5.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python
student_totals = []

for student in range(1, 4):
    print(f"Enter marks for Student {student}:")
    total = 0

    for subject in range(1, 4):
        while True:
            try:
                mark = float(input(f"Subject {subject}: "))
                if 0 <= mark <= 100:
                    total += mark
                    break
                print("Marks must be between 0 and 100.")
            except ValueError:
                print("Please enter a number.")

    student_totals.append(total)

print("Student totals:", student_totals)
print("Class average of totals:", sum(student_totals) / len(student_totals))
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
daily_totals = []

for day in range(1, 6):
    print(f"Day {day}:")
    daily_total = 0

    for product in range(1, 4):
        while True:
            try:
                sale = float(input(f"Product {product} sales amount: "))
                if sale < 0:
                    print("Sales cannot be negative.")
                    continue
                break
            except ValueError:
                print("Please enter a valid amount.")

        if sale == 0:
            print("No sale for this product; skipped.")
            continue

        daily_total += sale

    daily_totals.append(daily_total)
    print(f"Day {day} total: {daily_total:g}\n")

print("Daily Totals =", daily_totals)
print("Grand Total =", sum(daily_totals))
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python
for floor in range(1, 4):
    booked = 0
    available = 0
    print(f"Floor {floor}:")

    for room in range(1, 5):
        while True:
            answer = input(f"Room {room} booked? (Y/N): ").strip().upper()
            if answer == "Y":
                booked += 1
                break
            elif answer == "N":
                available += 1
                break
            print("Please enter Y or N.")

    print(f"Floor {floor}: {booked} booked, {available} available")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []

for invoice_number in range(1, 4):
    print(f"Invoice {invoice_number}:")
    items = []

    for item_number in range(1, 5):
        while True:
            try:
                price = float(input(f"Item {item_number} price: "))
                if price < 0:
                    print("Price cannot be negative.")
                    continue
                items.append(price)
                break
            except ValueError:
                print("Please enter a valid price.")

    subtotal = sum(items)
    tax = subtotal * 0.10
    total = subtotal + tax
    invoice_totals.append(total)
    print(f"Items: {items}")
    print(f"Subtotal: {subtotal:.2f}; Tax: {tax:.2f}; Total: {total:.2f}")

print("Invoice totals including tax:", invoice_totals)
```

```python
for student in range(1, 6):
    print(f"Student {student}:")
    marks = []
    failed_subjects = []

    for subject in range(1, 5):
        while True:
            try:
                mark = float(input(f"Subject {subject} marks: "))
                if 0 <= mark <= 100:
                    marks.append(mark)
                    if mark < 33:
                        failed_subjects.append(subject)
                    break
                print("Marks must be between 0 and 100.")
            except ValueError:
                print("Please enter a number.")

    total = sum(marks)
    average = total / len(marks)

    if failed_subjects:
        print("Failed in subject(s):", failed_subjects)
        grade = "F"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "E"

    print("Marks:", marks)
    print(f"Total: {total:g}/400; Average: {average:.2f}; Grade: {grade}\n")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
for package_number in range(1, 4):
    print(f"Package {package_number}:")
    ratings = []

    for feature in ["Hotel", "Transport", "Food"]:
        while True:
            try:
                rating = float(input(f"{feature} rating (0-5; 0 to skip): "))
                if 0 <= rating <= 5:
                    break
                print("Rating must be between 0 and 5.")
            except ValueError:
                print("Please enter a number.")

        if rating == 0:
            continue
        ratings.append(rating)

    if ratings:
        print(f"Ratings: {ratings} -> Average = {sum(ratings) / len(ratings):.2f}\n")
    else:
        print("No ratings provided for this package.\n")
```
