## A student is buying a laptop.

    Laptop price: Rs. 85,000
    Student has: Rs. 100,000
    Student gets a discount of Rs. 5,000
    A laptop bag costs Rs. 2,500
    The store has a promotion code "SAVE10".

### Create variables and:

    Calculate the laptop price after discount.
    Add the bag price to get the final bill.
    Calculate the remaining money.
    Check whether the student has enough money.
    Check whether "SAVE10" is present in the promotion code.
    Check whether the final bill is greater than Rs. 80,000.
    Store the result of whether the student has enough money in a variable and check whether it is truthy.

```python
laptop_price = 85000
available_money = 100000
discount = 5000
bag_price = 2500
promotion_code = "SAVE10"

discounted_price = laptop_price - discount
final_bill = discounted_price + bag_price
remaining_money = available_money - final_bill
enough_money = available_money >= final_bill
save10_present = "SAVE10" in promotion_code
bill_greater_than_80000 = final_bill > 80000
truthy_enough_money = bool(enough_money)

print("Laptop price after discount:", discounted_price)
print("Final bill:", final_bill)
print("Remaining money:", remaining_money)
print("Enough money:", enough_money)
print("SAVE10 present:", save10_present)
print("Final bill greater than 80000:", bill_greater_than_80000)
print("Enough money is truthy:", truthy_enough_money)
```

### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
math_marks = 85
english_marks = 72
python_marks = 90
total_possible_per_subject = 100
student_name = "Ahmed"
passing_marks = 50

total_marks = math_marks + english_marks + python_marks
average_marks = total_marks / 3
percentage = (total_marks / (total_possible_per_subject * 3)) * 100
passed_math = math_marks >= passing_marks
passed_all = (math_marks >= passing_marks and
              english_marks >= passing_marks and
              python_marks >= passing_marks)
name_contains_A = "A" in student_name
average_greater_equal_80 = average_marks >= 80
final_percentage = percentage
percentage_truthy = bool(final_percentage)

print("Total marks:", total_marks)
print("Average marks:", average_marks)
print("Percentage:", percentage)
print("Passed Mathematics:", passed_math)
print("Passed all subjects:", passed_all)
print('Name contains "A":', name_contains_A)
print("Average >= 80:", average_greater_equal_80)
print("Final percentage is truthy:", percentage_truthy)
```
