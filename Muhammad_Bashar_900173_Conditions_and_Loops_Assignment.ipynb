## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in metres: "))

if height <= 0 or weight < 0:
    print("Invalid weight or height")
else:
    bmi = weight / (height * height)
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    print(f"BMI = {bmi:.2f} → {category}")
```

### Goal:
    Help a shopper find the lowest price of the same product from three shops.

    Ask for the price in Shop 1, Shop 2 and Shop 3.
    
    Use nested if statements to find the smallest price (do not use min()).
    
    Print the lowest price and the shop number.

### Example:

    Enter prices: 1450, 1200, 1380
    Lowest price: 1200 (Shop 2)

```python
price1 = float(input("Enter price in Shop 1: "))
price2 = float(input("Enter price in Shop 2: "))
price3 = float(input("Enter price in Shop 3: "))

if price1 <= price2:
    if price1 <= price3:
        lowest, shop = price1, 1
    else:
        lowest, shop = price3, 3
else:
    if price2 <= price3:
        lowest, shop = price2, 2
    else:
        lowest, shop = price3, 3

print(f"Lowest price: {lowest:g} (Shop {shop})")
```

## ✊ 17. Rock, Paper, Scissors Game

### Goal:
    Decide the winner of a two-player game.

### Steps:

    Ask Player 1 and Player 2 for rock, paper or scissors.
    
    Rock beats scissors, scissors beats paper, paper beats rock.
    
    Use logical operators (and / or) to print the winner.
    
    If both choose the same → "Draw".

### Example:

    Player 1: rock
    Player 2: scissors
    Player 1 wins

```python
player1 = input("Player 1 (rock/paper/scissors): ").strip().lower()
player2 = input("Player 2 (rock/paper/scissors): ").strip().lower()

if player1 not in ("rock", "paper", "scissors") or player2 not in ("rock", "paper", "scissors"):
    print("Invalid choice")
elif player1 == player2:
    print("Draw")
elif (player1 == "rock" and player2 == "scissors") or (player1 == "scissors" and player2 == "paper") or (player1 == "paper" and player2 == "rock"):
    print("Player 1 wins")
else:
    print("Player 2 wins")
```

    Create settings = {"theme": "dark", "volume": 0, "nickname": ""}.

    Use settings.get(key) with a truthiness check.

    Add a comment explaining why volume (0) is treated as empty.

### Example:

    theme is set to dark

    nickname is empty or missing
    language is empty or missing

```python
settings = {"theme": "dark", "volume": 0, "nickname": ""}
keys = ["theme", "volume", "nickname", "language"]

for key in keys:
    value = settings.get(key)
    if value:
        print(f"{key} is set to {value}")
    else:
        # Zero is falsy in Python, so volume=0 is treated as empty here.
        print(f"{key} is empty or missing")
```

### Goal:

### Steps:

    Ask for the length and the width.

    The answer is the final value of a (the GCD).
    
    Print the tile size.

### Example:

    Length: 48
    Width: 18
    Largest tile = 6 ft

```python
a = int(input("Length: "))
b = int(input("Width: "))

a, b = abs(a), abs(b)
while b != 0:
    a, b = b, a % b

print(f"Largest tile = {a} ft")
```
