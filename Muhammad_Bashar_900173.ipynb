## 11. Student Report Card

**Scenario:** A school prints a report card from a dictionary of subject marks.

### Write a function `print_report_card(name, marks)` that:

* marks is a dictionary: subject name → marks

* Prints one line per subject in this format: <subject>: <marks>

* Average = sum of marks / number of subjects, rounded to 1 decimal

* Grade: 90 or more → A, 80–89 → B, 70–79 → C, 60–69 → D, below 60 → F

* Result is "Pass" if the average is 50 or more, otherwise "Fail"

* Finally prints: Average: <avg>, Grade: <grade> and Result: <result> (3 separate lines)

### 📥 Example Input:
    print_report_card("Ali", {"Math": 85, "Science": 90, "English": 78})
    print_report_card("Omar", {"Math": 40, "Science": 35, "English": 50})

### 📝 Example Output:
    Student: Ali
    Math: 85
    Science: 90
    English: 78
    Average: 84.3
    Grade: B
    Result: Pass
    
    Student: Omar
    Math: 40
    Science: 35
    English: 50
    Average: 41.7
    Grade: F
    Result: Fail

````python
def print_report_card(name, marks):
    print(f"Student: {name}")
    for subject, score in marks.items():
        print(f"{subject}: {score}")
    
    avg = round(sum(marks.values()) / len(marks), 1)
    
    if avg >= 90:
        grade = "A"
    elif avg >= 80:
        grade = "B"
    elif avg >= 70:
        grade = "C"
    elif avg >= 60:
        grade = "D"
    else:
        grade = "F"
        
    result = "Pass" if avg >= 50 else "Fail"
    
    print(f"Average: {avg}")
    print(f"Grade: {grade}")
    print(f"Result: {result}")

print_report_card("Ali", {"Math": 85, "Science": 90, "English": 78})
print_report_card("Omar", {"Math": 40, "Science": 35, "English": 50})
````

* Split it into words using spaces

## 22. Courier Charges

**Scenario:** A courier company charges by parcel weight.

### Write a function `courier_charge(weight_kg, is_express)` that:

* First 1 kg (or less) → 200 PKR

* Every extra kg → 80 PKR (a partial extra kg counts as a full kg, e.g. 2.5 extra kg = 3 kg)

* If is_express is True, add 50% to the total

* Returns the final charge

### 📥 Example Input:
    courier_charge(0.5, False)
    courier_charge(3.5, False)
    courier_charge(3.5, True)
    courier_charge(1, True)

### 📝 Example Output:
    200
    440
    660.0
    300.0

````python
import math

def courier_charge(weight_kg, is_express):
    base_charge = 200
    if weight_kg > 1:
        extra_kg = math.ceil(weight_kg - 1)
        total = base_charge + (extra_kg * 80)
    else:
        total = base_charge
        
    if is_express:
        total = total * 1.5
        
    return total

print(courier_charge(0.5, False))
print(courier_charge(3.5, False))
print(courier_charge(3.5, True))
print(courier_charge(1, True))
````

## 23. Gym Membership Price

**Scenario:** A gym charges 3000 PKR per month and offers layered discounts.

### Write a function `gym_membership_price(months, age, is_student)` that:

* Base price = 3000 × months

* Duration discount (only one applies): 12 months or more → 20% off, otherwise 6 months or more → 10% off

* Then apply an extra discount on the reduced price (only one applies): age 60 or more → 25% off, otherwise if is_student → 15% off

* Returns the final price rounded to 2 decimals

### 📥 Example Input:
    gym_membership_price(3, 30, False)
    gym_membership_price(12, 21, True)
    gym_membership_price(6, 65, True)

### 📝 Example Output:
    9000
    24480.0
    12150.0

````python
def gym_membership_price(months, age, is_student):
    base_price = 3000 * months
    
    # Duration discount
    if months >= 12:
        price_after_duration = base_price * 0.80
    elif months >= 6:
        price_after_duration = base_price * 0.90
    else:
        price_after_duration = base_price
        
    # Extra discount
    if age >= 60:
        final_price = price_after_duration * 0.75
    elif is_student:
        final_price = price_after_duration * 0.85
    else:
        final_price = price_after_duration
        
    return round(final_price, 2)

print(gym_membership_price(3, 30, False))
print(gym_membership_price(12, 21, True))
print(gym_membership_price(6, 65, True))
````

## 24. ATM Withdrawal

**Scenario:** An ATM must validate a withdrawal before giving cash.

### Write a function `atm_withdraw(balance, amount, withdrawn_today)` that:

* Check the rules in this order:

* 1. If amount is 0 or less, or not a multiple of 500 → return "Invalid Amount"

* 2. If withdrawn_today + amount is more than 50000 → return "Daily Limit Exceeded"

* 3. If amount is more than balance → return "Insufficient Balance"

* Otherwise return the new balance after withdrawing

### 📥 Example Input:
    atm_withdraw(20000, 5000, 0)
    atm_withdraw(20000, 1200, 0)
    atm_withdraw(100000, 30000, 25000)
    atm_withdraw(3000, 5000, 0)

### 📝 Example Output:

    'Invalid Amount'
    'Daily Limit Exceeded'
    'Insufficient Balance'

````python
def atm_withdraw(balance, amount, withdrawn_today):
    if amount <= 0 or amount % 500 != 0:
        return "Invalid Amount"
    if withdrawn_today + amount > 50000:
        return "Daily Limit Exceeded"
    if amount > balance:
        return "Insufficient Balance"
    return balance - amount

print(atm_withdraw(20000, 5000, 0))
print(atm_withdraw(20000, 1200, 0))
print(atm_withdraw(100000, 30000, 25000))
print(atm_withdraw(3000, 5000, 0))
````

## 25. Shop Stock Report

**Scenario:** A grocery shop updates its stock after a day of sales and finds what needs restocking.

### Write a function `restock_report(stock, sales, threshold)` that:

* stock is a dictionary: item name → quantity in stock

* sales is a list of dictionaries with keys "item" and "qty"

* For each sale: if the item is not in stock, or qty is more than the quantity in stock, the sale FAILS (add the item name to the failed list and do not change the stock). Otherwise subtract qty from the stock

* Do not modify the original stock dictionary — work on a copy

* Returns a dictionary with keys:

* "updated_stock" → the stock after all valid sales

* "low_stock" → alphabetical list of items whose updated quantity is below threshold

* "failed_sales" → list of item names whose sales failed, in the order they appeared

### 📥 Example Input:
    shop_stock = {"Rice": 50, "Oil": 20, "Sugar": 15, "Tea": 8}
    todays_sales = [
        {"item": "Rice", "qty": 10},
        {"item": "Oil", "qty": 25},
        {"item": "Sugar", "qty": 10},
        {"item": "Salt", "qty": 2},
    ]
    restock_report(shop_stock, todays_sales, 10)

### 📝 Example Output:
    {'updated_stock': {'Rice': 40, 'Oil': 20, 'Sugar': 5, 'Tea': 8},
     'low_stock': ['Sugar', 'Tea'],
     'failed_sales': ['Oil', 'Salt']}

````python
def restock_report(stock, sales, threshold):
    updated_stock = stock.copy()
    failed_sales = []
    
    for sale in sales:
        item = sale["item"]
        qty = sale["qty"]
        if item not in updated_stock or qty > updated_stock[item]:
            failed_sales.append(item)
        else:
            updated_stock[item] -= qty
            
    low_stock = sorted([item for item, qty in updated_stock.items() if qty < threshold])
    
    return {
        "updated_stock": updated_stock,
        "low_stock": low_stock,
        "failed_sales": failed_sales
    }

shop_stock = {"Rice": 50, "Oil": 20, "Sugar": 15, "Tea": 8}
todays_sales = [
    {"item": "Rice", "qty": 10},
    {"item": "Oil", "qty": 25},
    {"item": "Sugar", "qty": 10},
    {"item": "Salt", "qty": 2},
]
print(restock_report(shop_stock, todays_sales, 10))
````

* `create_account(owner, balance)` → returns a dictionary with keys "owner", "balance" and "history" (an empty list)

* `deposit(account, amount)` → if amount is 0 or less return "Invalid Amount". Otherwise add it to the balance, add "Deposit: <amount>" to the history and return "Deposit Successful"

* `withdraw(account, amount)` → if amount is 0 or less return "Invalid Amount". If amount is more than the balance return "Insufficient Balance". Otherwise subtract it, add "Withdraw: <amount>" to the history and return "Withdraw Successful"

* `transfer(sender, receiver, amount)` → a transfer has a 1% fee (fee = amount × 0.01, rounded to 2 decimals). If amount is 0 or less return "Invalid Amount". If amount + fee is more than the sender's balance return "Transfer Failed". Otherwise take amount + fee from the sender, give amount to the receiver, add "Transfer to <receiver name>: <amount> (Fee: <fee>)" to the sender's history, add "Transfer from <sender name>: <amount>" to the receiver's history and return "Transfer Successful"

* `print_statement(account)` → prints the owner, the balance, then "History:" followed by the numbered history lines (1. ..., 2. ...)

### ▶️ Then run this main program:

```python
ali = create_account("Ali", 10000)
sara = create_account("Sara", 5000)

print(deposit(ali, 5000))
print(withdraw(ali, 20000))
print(withdraw(ali, 3000))
print(transfer(ali, sara, 2000))
print(deposit(sara, -100))

print_statement(ali)
print_statement(sara)
```

### 📝 Expected Output:
    Deposit Successful
    Insufficient Balance
    Withdraw Successful
    Transfer Successful
    Invalid Amount
    Owner: Ali
    Balance: 9980.0
    History:
    1. Deposit: 5000
    2. Withdraw: 3000
    3. Transfer to Sara: 2000 (Fee: 20.0)
    Owner: Sara
    Balance: 7000
    History:
    1. Transfer from Ali: 2000

````python
def create_account(owner, balance):
    return {"owner": owner, "balance": balance, "history": []}

def deposit(account, amount):
    if amount <= 0:
        return "Invalid Amount"
    account["balance"] += amount
    account["history"].append(f"Deposit: {amount}")
    return "Deposit Successful"

def withdraw(account, amount):
    if amount <= 0:
        return "Invalid Amount"
    if amount > account["balance"]:
        return "Insufficient Balance"
    account["balance"] -= amount
    account["history"].append(f"Withdraw: {amount}")
    return "Withdraw Successful"

def transfer(sender, receiver, amount):
    if amount <= 0:
        return "Invalid Amount"
    fee = round(amount * 0.01, 2)
    if amount + fee > sender["balance"]:
        return "Transfer Failed"
    sender["balance"] -= (amount + fee)
    receiver["balance"] += amount
    sender["history"].append(f"Transfer to {receiver['owner']}: {amount} (Fee: {fee})")
    receiver["history"].append(f"Transfer from {sender['owner']}: {amount}")
    return "Transfer Successful"

def print_statement(account):
    print(f"Owner: {account['owner']}")
    print(f"Balance: {account['balance']}")
    print("History:")
    for idx, item in enumerate(account["history"], 1):
        print(f"{idx}. {item}")
````

````python
ali = create_account("Ali", 10000)
sara = create_account("Sara", 5000)

print(deposit(ali, 5000))
print(withdraw(ali, 20000))
print(withdraw(ali, 3000))
print(transfer(ali, sara, 2000))
print(deposit(sara, -100))

print_statement(ali)
print_statement(sara)
````

* `remove_item(cart, name)` → if the item is not in the cart return "Item not found"; otherwise delete it and return "Item removed"

* `calculate_subtotal(cart)` → returns the sum of price × qty for all items

* `get_discount(subtotal, coupon)` → coupon "SAVE10" gives 10% of the subtotal; coupon "FLAT500" gives 500 only if the subtotal is 3000 or more; any other coupon gives 0. Returns the discount amount rounded to 2 decimals

* `print_invoice(cart, coupon)` → prints one line per item in the format `Name — Qty: X — Total: Y`, then Subtotal, Discount, Tax (5% of subtotal minus discount, rounded to 2 decimals) and Final Amount (subtotal − discount + tax, rounded to 2 decimals)

### ▶️ Then run this main program:

```python

add_item(cart, "Shoes", 3000, 1)
```

    Discount: 400.0

````python
def add_item(cart, name, price, qty):
    if name in cart:
        cart[name]["qty"] += qty
    else:
        cart[name] = {"price": price, "qty": qty}
    return cart

def remove_item(cart, name):
    if name not in cart:
        return "Item not found"
    del cart[name]
    return "Item removed"

def calculate_subtotal(cart):
    return sum(item["price"] * item["qty"] for item in cart.values())

def get_discount(subtotal, coupon):
    if coupon == "SAVE10":
        return round(subtotal * 0.10, 2)
    elif coupon == "FLAT500":
        if subtotal >= 3000:
            return 500.0
        else:
            return 0.0
    else:
        return 0.0

def print_invoice(cart, coupon):
    for name, item in cart.items():
        total = item["price"] * item["qty"]
        print(f"{name} — Qty: {item['qty']} — Total: {total}")
    
    subtotal = calculate_subtotal(cart)
    discount = get_discount(subtotal, coupon)
    tax = round(0.05 * (subtotal - discount), 2)
    final_amount = round(subtotal - discount + tax, 2)
    
    print(f"Subtotal: {subtotal}")
    print(f"Discount: {discount}")
    print(f"Tax (5%): {tax}")
    print(f"Final Amount: {final_amount}")

cart = {}

add_item(cart, "Shoes", 3000, 1)
add_item(cart, "Socks", 250, 3)
add_item(cart, "Cap", 600, 1)
add_item(cart, "Socks", 250, 1)

print(remove_item(cart, "Watch"))
print(remove_item(cart, "Cap"))

print_invoice(cart, "SAVE10")
````

## 28. Mini Project 3: Student Result System

**Scenario:** A school wants to process the exam marks of a whole class and print a result sheet with the topper and the pass rate.

### Write these functions:

* `calculate_average(marks)` → returns the average of a list of marks rounded to 1 decimal

* `get_grade(avg)` → 90 or more → "A", 80–89 → "B", 70–79 → "C", 60–69 → "D", below 60 → "F"

* `build_result(name, marks)` → returns a dictionary with keys "name", "average", "grade" and "status". Status is "Pass" only if EVERY subject mark is 40 or more, otherwise "Fail"

* `find_topper(results)` → results is a list of dictionaries made by build_result. Returns the name of the student with the highest average (the first one if there is a tie). If the list is empty return "No students"

* `print_result_sheet(results)` → if the list is empty print "No students" and stop. Otherwise print the title `--- Result Sheet ---`, one line per student in the format `Name | Avg: X | Grade: Y | Status`, then `Topper: <name>` and `Pass Rate: <percentage>%` (students who passed ÷ total students × 100, rounded to 1 decimal)

### ▶️ Then run this main program:

```python
students = [
    ("Ali", [85, 90, 78]),
    ("Sara", [95, 88, 92]),
    ("Omar", [40, 35, 50]),
    ("Hina", [72, 65, 80]),
]

results = []
for name, marks in students:
    results.append(build_result(name, marks))

print_result_sheet(results)
```

### 📝 Expected Output:
    --- Result Sheet ---
    Ali | Avg: 84.3 | Grade: B | Pass
    Sara | Avg: 91.7 | Grade: A | Pass
    Omar | Avg: 41.7 | Grade: F | Fail
    Hina | Avg: 72.3 | Grade: C | Pass
    Topper: Sara
    Pass Rate: 75.0%

````python
def calculate_average(marks):
    return round(sum(marks) / len(marks), 1)

def get_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"

def build_result(name, marks):
    avg = calculate_average(marks)
    grade = get_grade(avg)
    status = "Pass" if all(m >= 40 for m in marks) else "Fail"
    return {"name": name, "average": avg, "grade": grade, "status": status}

def find_topper(results):
    if not results:
        return "No students"
    topper = max(results, key=lambda x: x["average"])
    return topper["name"]

def print_result_sheet(results):
    if not results:
        print("No students")
        return
    
    print("--- Result Sheet ---")
    passed_count = 0
    for r in results:
        print(f"{r['name']} | Avg: {r['average']} | Grade: {r['grade']} | {r['status']}")
        if r['status'] == "Pass":
            passed_count += 1
            
    topper = find_topper(results)
    pass_rate = round((passed_count / len(results)) * 100, 1)
    
    print(f"Topper: {topper}")
    print(f"Pass Rate: {pass_rate}%")

students = [
    ("Ali", [85, 90, 78]),
    ("Sara", [95, 88, 92]),
    ("Omar", [40, 35, 50]),
    ("Hina", [72, 65, 80]),
]

results = []
for name, marks in students:
    results.append(build_result(name, marks))

print_result_sheet(results)
````

## 30. Mini Project 5: Employee Payroll System

**Scenario:** A company runs payroll at the end of every month: overtime, performance bonus, progressive income tax, insurance and a payslip for each employee.

### Write these functions:

* `calculate_overtime_pay(hours)` → 1 overtime hour = 300 PKR. Returns the overtime pay

* `calculate_bonus(basic, performance)` → "excellent" → 20% of basic, "good" → 10%, "average" → 5%, anything else → 0. Returns the bonus

* `calculate_tax(gross)` → progressive tax: up to 50000 → 0; the part between 50000 and 100000 → 5%; the part above 100000 → 12% (plus the full 5% on the 50000–100000 part). Returns the tax

* `generate_payslip(employee)` → the employee is a dictionary with keys "name", "basic", "overtime_hours" and "performance". Gross = basic + overtime pay + bonus. Medical insurance is a fixed 1000 PKR. Net = gross − tax − insurance. Returns a dictionary with keys "name", "overtime", "bonus", "gross", "tax", "insurance" and "net" (money values rounded to 2 decimals)

* `print_payslip(slip)` → prints the 7 lines: `Employee`, `Overtime Pay`, `Performance Bonus`, `Gross Salary`, `Income Tax`, `Medical Insurance`, `Net Salary`

* `run_payroll(employees)` → for every employee: generate the payslip, print it, then print a line of dashes `--------`. At the end print `Total Payroll: <sum of all net salaries>` (rounded to 2 decimals) and return that total

### ▶️ Then run this main program:

```python
employees = [
    {"name": "Ali", "basic": 50000, "overtime_hours": 20, "performance": "excellent"},
    {"name": "Sara", "basic": 90000, "overtime_hours": 10, "performance": "good"},
    {"name": "Omar", "basic": 30000, "overtime_hours": 0, "performance": "poor"},
]

run_payroll(employees)
```

### 📝 Expected Output:
    Employee: Ali
    Overtime Pay: 6000
    Performance Bonus: 10000.0
    Gross Salary: 66000.0
    Income Tax: 800.0
    Medical Insurance: 1000
    Net Salary: 64200.0
    --------
    Employee: Sara
    Overtime Pay: 3000
    Performance Bonus: 9000.0
    Gross Salary: 102000.0
    Income Tax: 2740.0
    Medical Insurance: 1000
    Net Salary: 98260.0
    --------
    Employee: Omar
    Overtime Pay: 0
    Performance Bonus: 0
    Gross Salary: 30000
    Income Tax: 0
    Medical Insurance: 1000
    Net Salary: 29000
    --------
    Total Payroll: 191460.0

````python
def calculate_overtime_pay(hours):
    return hours * 300

def calculate_bonus(basic, performance):
    if performance == "excellent":
        return basic * 0.20
    elif performance == "good":
        return basic * 0.10
    elif performance == "average":
        return basic * 0.05
    else:
        return 0

def calculate_tax(gross):
    if gross <= 50000:
        return 0
    elif gross <= 100000:
        return (gross - 50000) * 0.05
    else:
        return (50000 * 0.05) + ((gross - 100000) * 0.12)

def generate_payslip(employee):
    name = employee["name"]
    basic = employee["basic"]
    overtime_hours = employee["overtime_hours"]
    performance = employee["performance"]
    
    overtime = calculate_overtime_pay(overtime_hours)
    bonus = calculate_bonus(basic, performance)
    gross = basic + overtime + bonus
    tax = calculate_tax(gross)
    insurance = 1000
    net = gross - tax - insurance
    
    return {
        "name": name,
        "overtime": round(overtime, 2),
        "bonus": round(bonus, 2),
        "gross": round(gross, 2),
        "tax": round(tax, 2),
        "insurance": round(insurance, 2),
        "net": round(net, 2)
    }

def print_payslip(slip):
    print(f"Employee: {slip['name']}")
    print(f"Overtime Pay: {slip['overtime']}")
    print(f"Performance Bonus: {slip['bonus']}")
    print(f"Gross Salary: {slip['gross']}")
    print(f"Income Tax: {slip['tax']}")
    print(f"Medical Insurance: {slip['insurance']}")
    print(f"Net Salary: {slip['net']}")

def run_payroll(employees):
    total_net = 0
    for emp in employees:
        slip = generate_payslip(emp)
        print_payslip(slip)
        print("--------")
        total_net += slip["net"]
    total_net = round(total_net, 2)
    print(f"Total Payroll: {total_net}")
    return total_net
````

````python
employees = [
    {"name": "Ali", "basic": 50000, "overtime_hours": 20, "performance": "excellent"},
    {"name": "Sara", "basic": 90000, "overtime_hours": 10, "performance": "good"},
    {"name": "Omar", "basic": 30000, "overtime_hours": 0, "performance": "poor"},
]

run_payroll(employees)
````
